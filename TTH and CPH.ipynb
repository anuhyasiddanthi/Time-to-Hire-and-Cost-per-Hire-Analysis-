```python
# Year-wise average Time to Hire
time_year = df.groupby("Year")["Total_Days_Per_Candidate"].mean().reset_index()

# Prepare data for model
X = time_year["Year"].values.reshape(-1,1)
y = time_year["Total_Days_Per_Candidate"].values

# Train Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Predict future years
future_years = np.array([2025, 2026]).reshape(-1,1)
predicted = model.predict(future_years)

# Print predicted values
print("Average Predicted Time to Hire")
print("2025:", round(predicted[0],2), "days")
print("2026:", round(predicted[1],2), "days")

# Plot graph
plt.figure(figsize=(8,5))

# Actual values
plt.plot(time_year["Year"], 
         time_year["Total_Days_Per_Candidate"],
         marker='o',
         label="Actual")

# Predicted values
plt.plot([2025, 2026],
         predicted,
         marker='x',
         linestyle='--',
         color='red',
         label="Predicted")

plt.title("Avg Time to Hire Trend (2018–2026)")
plt.xlabel("Year")
plt.ylabel("Days")

plt.legend()
plt.grid(True)

plt.show()
```

```python
# Calculate year-wise average cost
cost_year = df.groupby("Year")["Total_Cost_Per_Candidate"].mean().reset_index()

# Prepare data
X = cost_year["Year"].values.reshape(-1,1)
y = cost_year["Total_Cost_Per_Candidate"].values

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict future years
future_years = np.array([2025, 2026]).reshape(-1,1)
predicted_cost = model.predict(future_years)

# Print predicted values
print("Predicted Average Cost per Hire in 2025:", round(predicted_cost[0],2))
print("Predicted Average Cost per Hire in 2026:", round(predicted_cost[1],2))

# Plot graph
plt.figure(figsize=(8,5))

# Actual values
plt.plot(cost_year["Year"],
         cost_year["Total_Cost_Per_Candidate"],
         marker='o',
         label="Actual")

# Predicted values
plt.plot([2025, 2026],
         predicted_cost,
         marker='x',
         linestyle='--',
         color='red',
         label="Predicted")

plt.title("Trend: Average Cost per Hire (2018–2026)")
plt.xlabel("Year")
plt.ylabel("Cost (INR)")

plt.legend()
plt.grid(True)

plt.show()
```

```python
# Create list to store predictions
future_data = []

# Predict for each job title
for job in df["Job_Title"].unique():

    job_df = df[df["Job_Title"] == job]

    X = job_df["Year"].values.reshape(-1,1)
    y = job_df["Total_Days_Per_Candidate"].values

    model = LinearRegression()
    model.fit(X,y)

    future_years = np.array([2025,2026]).reshape(-1,1)
    predictions = model.predict(future_years)

    future_data.append([2025, job, predictions[0]])
    future_data.append([2026, job, predictions[1]])

# Convert predictions to dataframe
future_df = pd.DataFrame(
    future_data,
    columns=["Year","Job_Title","Total_Days_Per_Candidate"]
)

# Combine actual + predicted data
df_full = pd.concat([df, future_df], ignore_index=True)

# Plot graph
plt.figure(figsize=(12,6))

sns.barplot(
    x="Year",
    y="Total_Days_Per_Candidate",
    hue="Job_Title",
    data=df_full,
    palette="tab20",
    ci=None
)

plt.title("Time to Hire Histogram by Job Title (2018–2026)")
plt.xlabel("Year")
plt.ylabel("Average Time to Hire (Days)")

plt.xticks(rotation=45)

plt.legend(bbox_to_anchor=(1.05,1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# Load your uploaded dataset
df = pd.read_csv(r"C:\Users\Dell\Downloads\TTH and CPH_3.csv")

plt.figure(figsize=(8,5))

for recruiter in df["Recruiter"].unique():

    rec_df = df[df["Recruiter"] == recruiter]

    X = rec_df["Year"].values.reshape(-1,1)
    y = rec_df["Avg_Time"].values

    model = LinearRegression()
    model.fit(X,y)

    future_years = np.array([2025,2026]).reshape(-1,1)
    pred = model.predict(future_years)

    # Plot actual line and capture its color
    line = plt.plot(rec_df["Year"], y, marker='o', label=recruiter)

    color = line[0].get_color()

    # Plot predicted line using the same color
    plt.plot([2025,2026], pred,
             marker='x',
             linestyle='--',
             color=color)

plt.title("Trend and Prediction of Average Time to Hire by Recruiter")
plt.xlabel("Year")
plt.ylabel("Average Time to Hire (Days)")

plt.legend(title="Recruiter")
plt.grid(True)

plt.show()
```
